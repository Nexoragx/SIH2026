```python
def generate_patient_text(row):

    texts = []

    if row["event_type"] == "Divorce":
        texts.append("I am struggling emotionally after my divorce.")

    elif row["event_type"] == "Death of family member":
        texts.append("I am finding it very difficult to cope with the loss of my family member.")

    elif row["event_type"] == "Unsuccessful exam attempt":
        texts.append("I failed my exam and I feel disappointed and worried about my future.")

    elif row["event_type"] == "Repeated exam failure":
        texts.append("I have failed several times and I feel hopeless about my studies.")

    elif row["event_type"] == "Job loss":
        texts.append("I lost my job and I am worried about what will happen next.")

    elif row["event_type"] == "Unable to get a job":
        texts.append("I have been trying to find a job without success and it is becoming very stressful.")

    elif row["event_type"] == "Legal dispute":
        texts.append("The ongoing legal dispute is causing me significant stress.")

    elif row["event_type"] == "Wrongful arrest":
        texts.append("I feel distressed and angry about being wrongly arrested.")

    elif row["event_type"] == "Arrest/legal proceedings":
        texts.append("The legal proceedings are causing me a lot of worry.")

    elif row["event_type"] == "Relationship breakup":
        texts.append("I am struggling to cope after my relationship ended.")

    elif row["event_type"] == "Financial crisis":
        texts.append("My financial situation is causing me constant worry.")

    elif row["event_type"] == "Academic rejection":
        texts.append("I was rejected academically and I feel discouraged.")

    elif row["event_type"] == "Business failure":
        texts.append("My business failed and I am struggling emotionally.")

    elif row["event_type"] == "Family conflict":
        texts.append("There is a lot of conflict in my family and I feel overwhelmed.")

    else:
        texts.append("I am going through a difficult period and I feel emotionally exhausted.")

    return texts[0]

df["text"] = df.apply(generate_patient_text, axis=1)

df[["event_type", "text"]].head(10)
```

```python
from transformers import pipeline

sentiment_model = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
text = "I feel hopeless and exhausted after losing my job."

sentiment_model(text)
```

```python
sentiment_results = sentiment_model(
    df["text"].tolist(),
    batch_size=32,
    truncation=True
)

df["sentiment"] = [
    r["label"] for r in sentiment_results
]

df["sentiment_confidence"] = [
    r["score"] for r in sentiment_results
]

df[["text", "sentiment", "sentiment_confidence"]].head()
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("synthetic_MADRS_patient_dataset_5000.csv")

df.head()
```

**Emotion Detection**

```python
emotion_model = pipeline(
    "text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    top_k=1
)
```

```python
emotion_results = emotion_model(
    df["text"].tolist(),
    batch_size=32,
    truncation=True
)

df["emotion"] = [
    r[0]["label"] for r in emotion_results
]

df["emotion_confidence"] = [
    r[0]["score"] for r in emotion_results
]

df[["text", "emotion", "emotion_confidence"]].head()
```

***Phase3***

```python
madrs_features = [
    "apparent_sadness",
    "reported_sadness",
    "inner_tension",
    "reduced_sleep",
    "reduced_appetite",
    "concentration_difficulties",
    "lassitude",
    "inability_to_feel",
    "pessimistic_thoughts",
    "suicidal_thoughts"
]

categorical_features = [
    "gender",
    "event_type",
    "event_status",
    "sentiment",
    "emotion"
]

numeric_features = [
    "age",
    "months_since_event",
    "sentiment_confidence",
    "emotion_confidence"
]

X = df[madrs_features + categorical_features + numeric_features]
y = df["severity_category"]
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=categorical_features,
    drop_first=True
)

X_encoded = X_encoded.astype(float)

X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

multimodal_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=12,
    min_samples_leaf=5,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

multimodal_model.fit(X_train, y_train)

print("Multimodal model trained successfully.")
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score
)

y_pred = multimodal_model.predict(X_test)

print("Accuracy:",
      accuracy_score(y_test, y_pred))

print("\nMacro F1:",
      f1_score(y_test, y_pred, average="macro"))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    xticks_rotation=45
)

plt.title("SahaayAI Multimodal Model")
plt.tight_layout()
plt.show()
```

```python
importance = pd.DataFrame({
    "feature": X_train.columns,
    "importance": multimodal_model.feature_importances_
})

importance = importance.sort_values(
    "importance",
    ascending=False
)

importance.head(20)
```

```python
plt.figure(figsize=(10, 7))

top = importance.head(15).sort_values("importance")

plt.barh(
    top["feature"],
    top["importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("SahaayAI Feature Importance")

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# Separate target variable (y) from features (X)
y = df['severity_category']
X = df.drop(columns=['severity_category', 'patient_id'])

# Identify categorical features for one-hot encoding
categorical_features = X.select_dtypes(include=['object']).columns
X_encoded = pd.get_dummies(X, columns=categorical_features, drop_first=True)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42, stratify=y)

# Display the shapes of the resulting datasets
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

## Train Random Forest Model

### Subtask:
Train a Random Forest classifier on the prepared training data (`X_train`, `y_train`). The target variable is 'severity_category'.


**Reasoning**:
Import `RandomForestClassifier` from `sklearn.ensemble`, instantiate it with `n_estimators=100` and `random_state=42` for reproducibility, and then train the model using the `.fit()` method on the training data (`X_train`, `y_train`).



```python
from sklearn.ensemble import RandomForestClassifier

# Instantiate a RandomForestClassifier model
# Using n_estimators=100 and random_state=42 for reproducibility
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model on the training data
model.fit(X_train, y_train)

print("Random Forest model trained successfully.")
```
